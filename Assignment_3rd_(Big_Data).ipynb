```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("pyspark_tutorial_4").getOrCreate()
```

```python
df = spark.read.csv("/content/drive/MyDrive/Datasets.xlsx - Dataset-Sales.csv", header = True, inferSchema =True)
```

```python
df.show(10)
```

```python
from pyspark.sql import functions as F

# 1. Filter operation - extract rows from DF based on conditions
# Filter for 'Foot Locker' retailers
df.filter(F.col("Retailer") == "Foot Locker").show(10)
```

```python
# Filter for 'Units sold'
df.filter(F.regexp_replace(F.col("Units Sold"), ",", "").cast("int") > 1000).show(10)
```

```python
# 2. Logical operators (AND and OR)
# Filter for 'Foot Locker' retailers AND 'Units Sold' greater than 1000
df.filter((F.col("Retailer") == "Foot Locker") & (F.regexp_replace(F.col("Units Sold"), ",", "").cast("int") > 1000)).show(10)
```

```python
# Filter for 'Northeast' region OR 'West' region
df.filter((F.col("Region") == "Northeast") | (F.col("Region") == "West")).show(10)
```

```python
# 3. Comparison operators (== and !=)
# Filter for products where the Retailer is not 'Foot Locker'
df.filter(F.col("Retailer") != "Foot Locker").show(10)
```

```python
# Filter for products where the 'Sales Method' is 'Online'
df.filter(F.col("Sales Method") == "Online").show(10)
```

```python
# 4. Groupby
# Group by 'Retailer' and calculate the sum of 'Total Sales'
df.groupBy("Retailer").agg(F.sum(F.regexp_replace(F.col("Total Sales"), "[$,]", "").cast("double")).alias("Total Sales Sum")).show()
```

```python
# 5. Overall Aggregated values function
# Calculate the overall sum of 'Units Sold' and 'Total Sales'
df.agg(
    F.sum(F.regexp_replace(F.col("Units Sold"), ",", "").cast("int")).alias("Overall Units Sold"),
    F.sum(F.regexp_replace(F.col("Total Sales"), "[$,]", "").cast("double")).alias("Overall Total Sales")
).show()
```

```python
# 6. format_number, Alias
# Format 'Total Sales' to 2 decimal places and rename the column
df.select(
    F.format_number(F.regexp_replace(F.col("Total Sales"), "[$,]", "").cast("double"), 2).alias("Formatted Total Sales"),
    F.format_number(F.regexp_replace(F.col("Operating Profit"), "[$,]", "").cast("double"), 2).alias("Formatted Operating Profit")
).show(10)
```

```python
# 7. Sort data using orderBy
# Order the DataFrame by 'Total Sales' in descending order
df.orderBy(F.regexp_replace(F.col("Total Sales"), "[$,]", "").cast("double").desc()).show(10)

```

```python
# Order by 'Retailer' (ascending) and then 'Units Sold' (descending)
df.orderBy(F.col("Retailer").asc(), F.regexp_replace(F.col("Units Sold"), ",", "").cast("int").desc()).show(10)
```
