```python
#srart spark session
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("select_ops").getOrCreate()
```

```python
# connect to the drive
df = spark.read.option('header',True).option('inferschema',True).csv("/content/sample_data/california_housing_test.csv")
```

```python
# read the dataset
df.show(10)
```

```python
# check the datatype of the dataframe
df.printSchema()
```

```python
# check the sumary
df.describe().show()
```

```python
df.summary().show()
```

```python
#select the columns
df.select(["longitude","housing_median_age","total_rooms","population"]).show()
```

```python
#add new columns
df_rooms_per_household= df.withColumn("rooms_per_household" , df['total_rooms'] / df['households'])
df_rooms_per_household.show(5)
```

```python
#drop the columns
df_final_columns = df.drop('longitude','latitude')
df_final_columns.show(5)
```

```python
# Rename columns
df = df.withColumnRenamed("median_house_value", "house_value").withColumnRenamed("housing_median_age", "house_age")
df.show(5)
```
